```python
# Cell 15: Calculate daily returns and Garman-Klass variance

import numpy as np
import pandas as pd

market_df = market_df.sort_values(
    ["Symbol", "Date"]
).reset_index(drop=True)

# Previous trading day's adjusted close for each stock
market_df["Prev_Adj_Close"] = (
    market_df.groupby("Symbol")["Adj Close"].shift(1)
)

# Daily log return using adjusted closing prices
market_df["log_return"] = np.log(
    market_df["Adj Close"] / market_df["Prev_Adj_Close"]
)

# Garman-Klass daily variance estimator
log_hl = np.log(market_df["High"] / market_df["Low"])
log_co = np.log(market_df["Close"] / market_df["Open"])

market_df["gk_variance"] = (
    0.5 * log_hl**2
    - (2 * np.log(2) - 1) * log_co**2
)

# Numerical check: Garman-Klass estimates can be negative
# for some individual OHLC configurations, so inspect before deciding
# how to construct the final target.
print("RETURN AND GARMAN-KLASS SUMMARY")
print("-" * 50)
print("Valid log returns:", market_df["log_return"].notna().sum())
print("Missing log returns:", market_df["log_return"].isna().sum())
print("Negative GK estimates:", (market_df["gk_variance"] < 0).sum())
print("Zero GK estimates:", (market_df["gk_variance"] == 0).sum())

print("\nDescriptive statistics:")
display(market_df[["log_return", "gk_variance"]].describe().T)

print("\nSample observations:")
display(
    market_df[
        ["Date", "Symbol", "Open", "High", "Low", "Close",
         "Adj Close", "log_return", "gk_variance"]
    ].head(10)
)
```

```python
# Cell 16: Create future volatility targets

import numpy as np
import pandas as pd

market_df = market_df.sort_values(
    ["Symbol", "Date"]
).reset_index(drop=True)

grouped_gk = market_df.groupby("Symbol")["gk_variance"]

# Target 1: variance on the next trading observation
market_df["target_gk_var_next1d"] = grouped_gk.shift(-1)

# Target 2: average variance across the next five trading observations.
# Reverse the series so the rolling window represents future observations.
market_df["target_gk_var_next5d"] = (
    grouped_gk.transform(
        lambda s: s.shift(-1).iloc[::-1]
        .rolling(window=5, min_periods=5)
        .mean()
        .iloc[::-1]
    )
)

# Log targets; floor is used only to handle zero variance values
VARIANCE_FLOOR = 1e-12

market_df["target_log_var_next1d"] = np.log(
    market_df["target_gk_var_next1d"].clip(lower=VARIANCE_FLOOR)
)

market_df["target_log_var_next5d"] = np.log(
    market_df["target_gk_var_next5d"].clip(lower=VARIANCE_FLOOR)
)

print("FUTURE VOLATILITY TARGETS")
print("-" * 50)

target_cols = [
    "target_gk_var_next1d",
    "target_gk_var_next5d",
    "target_log_var_next1d",
    "target_log_var_next5d",
]

display(market_df[target_cols].describe().T)

print("\nMissing targets:")
display(market_df[target_cols].isna().sum().to_frame("missing_count"))

print("\nSample rows:")
display(
    market_df[
        ["Date", "Symbol", "gk_variance"] + target_cols
    ].head(10)
)
```

```python
# Cell 17: Verify that future targets are correctly aligned

import numpy as np
import pandas as pd

market_df = market_df.sort_values(
    ["Symbol", "Date"]
).reset_index(drop=True)

# Verify next-day target equals the next row's GK variance
expected_next1d = (
    market_df.groupby("Symbol")["gk_variance"].shift(-1)
)

next1d_match = np.isclose(
    market_df["target_gk_var_next1d"],
    expected_next1d,
    equal_nan=True
)

# Independently calculate the mean of the next five observations
expected_next5d = market_df.groupby("Symbol")["gk_variance"].transform(
    lambda s: s.shift(-1).iloc[::-1]
    .rolling(window=5, min_periods=5)
    .mean()
    .iloc[::-1]
)

next5d_match = np.isclose(
    market_df["target_gk_var_next5d"],
    expected_next5d,
    equal_nan=True
)

print("TARGET ALIGNMENT VALIDATION")
print("-" * 50)
print(
    "Next-day target matches expected:",
    int(next1d_match.sum()), "/", len(market_df)
)
print(
    "Next-five-day target matches expected:",
    int(next5d_match.sum()), "/", len(market_df)
)

print("\nFirst stock's final six observations:")
display(
    market_df.loc[
        market_df["Symbol"] == market_df["Symbol"].iloc[0],
        [
            "Date", "Symbol", "gk_variance",
            "target_gk_var_next1d", "target_gk_var_next5d"
        ]
    ].tail(6)
)
```

```python
# Cell 18: Create historical volatility predictors

import numpy as np
import pandas as pd

market_df = market_df.sort_values(
    ["Symbol", "Date"]
).reset_index(drop=True)

grouped = market_df.groupby("Symbol", group_keys=False)

# Lagged daily variance: information available before today's close
market_df["gk_variance_lag1"] = (
    grouped["gk_variance"].shift(1)
)

# Rolling mean variance, using only observations up to the current day
for window in [5, 10, 22]:
    market_df[f"gk_mean_{window}d"] = (
        grouped["gk_variance"]
        .transform(
            lambda s: s.rolling(window=window, min_periods=window).mean()
        )
    )

# Rolling volatility of daily log returns
for window in [5, 22]:
    market_df[f"return_volatility_{window}d"] = (
        grouped["log_return"]
        .transform(
            lambda s: s.rolling(window=window, min_periods=window).std()
        )
    )

feature_cols = [
    "gk_variance_lag1",
    "gk_mean_5d",
    "gk_mean_10d",
    "gk_mean_22d",
    "return_volatility_5d",
    "return_volatility_22d",
]

print("HISTORICAL VOLATILITY FEATURES")
print("-" * 50)
display(market_df[feature_cols].describe().T)

print("\nMissing values in features:")
display(market_df[feature_cols].isna().sum().to_frame("missing_count"))

print("\nSample rows:")
display(
    market_df[
        ["Date", "Symbol", "gk_variance", "log_return"] + feature_cols
    ].head(30)
)
```

```python
# Cell 21: Merge market features with sentiment data

import pandas as pd

# Load the saved market features
market_features_path = FEATURES_DIR / "market_volatility_features.parquet"
market_features_df = pd.read_parquet(market_features_path)

# Standardize market symbols: RELIANCE.NS -> RELIANCE
market_features_df["Symbol"] = (
    market_features_df["Symbol"]
    .astype("string")
    .str.strip()
    .str.upper()
    .str.replace(r"\.NS$", "", regex=True)
)

# Ensure both datasets use the same date format
market_features_df["Date"] = pd.to_datetime(
    market_features_df["Date"]
).dt.normalize()

sentiment_df["Date"] = pd.to_datetime(
    sentiment_df["Date"]
).dt.normalize()

# Identify sentiment columns to merge
sentiment_feature_cols = [
    col for col in sentiment_df.columns
    if col not in ["Date", "Symbol"]
]

# Merge while preserving all market observations
merged_df = market_features_df.merge(
    sentiment_df,
    on=["Date", "Symbol"],
    how="left",
    validate="one_to_one",
    indicator=True
)

# Summarize merge coverage
matched_rows = (merged_df["_merge"] == "both").sum()
unmatched_rows = (merged_df["_merge"] == "left_only").sum()

print("MARKET + SENTIMENT MERGE")
print("-" * 50)
print("Market rows:", len(market_features_df))
print("Merged rows:", len(merged_df))
print("Matched rows:", matched_rows)
print("Unmatched rows:", unmatched_rows)
print("Match rate:", f"{matched_rows / len(merged_df):.2%}")

# Missing sentiment features after the merge
print("\nMissing sentiment values after merge:")
display(
    merged_df[sentiment_feature_cols]
    .isna()
    .sum()
    .sort_values(ascending=False)
    .to_frame("missing_count")
    .head(15)
)

# Save merge diagnostics separately from model features
merge_counts = merged_df["_merge"].value_counts()
print("\nMerge status counts:")
display(merge_counts.to_frame("row_count"))

# Save merged dataset
merged_df = merged_df.drop(columns="_merge")

merged_csv_path = FEATURES_DIR / "market_sentiment_merged.csv"
merged_parquet_path = FEATURES_DIR / "market_sentiment_merged.parquet"

merged_df.to_csv(merged_csv_path, index=False)
merged_df.to_parquet(merged_parquet_path, index=False)

print("\nSaved merged dataset:")
print(merged_parquet_path.resolve())
print("Total columns:", len(merged_df.columns))
```

```python
# Cell 30: Train HAR-style historical-volatility benchmark

import pandas as pd
import numpy as np
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score
import joblib

# Historical-only HAR predictors
har_features = [
    "gk_variance_lag1",
    "gk_mean_5d",
    "gk_mean_22d",
]

horizons = {
    "1d": "target_log_var_next1d",
    "5d": "target_log_var_next5d",
}

har_models = {}
har_results = []

for horizon, target_col in horizons.items():
    train = pd.read_parquet(
        FEATURES_DIR / f"train_{horizon}_model.parquet"
    )
    validation = pd.read_parquet(
        FEATURES_DIR / f"validation_{horizon}_model.parquet"
    )

    # Use the same features for both horizons
    X_train = train[har_features].replace(
        [np.inf, -np.inf], np.nan
    )
    y_train = train[target_col]

    X_validation = validation[har_features].replace(
        [np.inf, -np.inf], np.nan
    )
    y_validation = validation[target_col]

    # Keep only finite rows
    train_mask = (
        np.isfinite(X_train).all(axis=1)
        & np.isfinite(y_train)
    )
    validation_mask = (
        np.isfinite(X_validation).all(axis=1)
        & np.isfinite(y_validation)
    )

    X_train = X_train.loc[train_mask]
    y_train = y_train.loc[train_mask]
    X_validation = X_validation.loc[validation_mask]
    y_validation = y_validation.loc[validation_mask]

    model = LinearRegression()
    model.fit(X_train, y_train)

    predictions = model.predict(X_validation)

    har_models[horizon] = model

    har_results.append({
        "Horizon": horizon,
        "Train rows": len(X_train),
        "Validation rows": len(X_validation),
        "Validation RMSE": np.sqrt(
            mean_squared_error(y_validation, predictions)
        ),
        "Validation MAE": mean_absolute_error(
            y_validation, predictions
        ),
        "Validation R2": r2_score(
            y_validation, predictions
        ),
    })

    joblib.dump(
        model,
        MODELS_DIR / f"har_{horizon}.joblib"
    )

har_results_df = pd.DataFrame(har_results)

print("HAR BENCHMARK — VALIDATION PERFORMANCE")
print("-" * 55)
print(har_results_df.round(6).to_string(index=False))
print("\nModels saved in:", MODELS_DIR)
```

```python
# Cell 35: LightGBM benchmark models

import pandas as pd
import numpy as np
import lightgbm as lgb
import joblib

from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score

historical_features = [
    "gk_variance_lag1",
    "gk_mean_5d",
    "gk_mean_10d",
    "gk_mean_22d",
    "return_volatility_5d",
    "return_volatility_22d",
]

news_features = [
    "direct_news_pos",
    "direct_news_neu",
    "direct_news_neg",
    "sectoral_news_pos",
    "sectoral_news_neu",
    "sectoral_news_neg",
    "global_news_pos",
    "global_news_neu",
    "global_news_neg",
    "direct_news_count",
    "sectoral_news_count",
    "global_news_count",
]

model_feature_sets = {
    "LGBM-Historical": historical_features,
    "LGBM-News": historical_features + news_features,
}

horizons = {
    "1d": "target_log_var_next1d",
    "5d": "target_log_var_next5d",
}

lgbm_models = {}
lgbm_results = []

for horizon, target_col in horizons.items():
    train = pd.read_parquet(
        FEATURES_DIR / f"train_{horizon}_model.parquet"
    )
    validation = pd.read_parquet(
        FEATURES_DIR / f"validation_{horizon}_model.parquet"
    )

    train = train.replace([np.inf, -np.inf], np.nan)
    validation = validation.replace([np.inf, -np.inf], np.nan)

    for model_name, feature_cols in model_feature_sets.items():
        train_model = train.dropna(
            subset=feature_cols + [target_col]
        )
        val_model = validation.dropna(
            subset=feature_cols + [target_col]
        )

        X_train = train_model[feature_cols]
        y_train = train_model[target_col]
        X_val = val_model[feature_cols]
        y_val = val_model[target_col]

        model = lgb.LGBMRegressor(
            objective="regression",
            n_estimators=500,
            learning_rate=0.03,
            num_leaves=31,
            max_depth=-1,
            subsample=0.8,
            colsample_bytree=0.8,
            reg_lambda=1.0,
            random_state=42,
            n_jobs=-1,
            verbosity=-1,
        )

        model.fit(X_train, y_train)
        predictions = model.predict(X_val)

        lgbm_models[(horizon, model_name)] = model

        lgbm_results.append({
            "Horizon": horizon,
            "Model": model_name,
            "Train rows": len(train_model),
            "Validation rows": len(val_model),
            "RMSE": np.sqrt(
                mean_squared_error(y_val, predictions)
            ),
            "MAE": mean_absolute_error(y_val, predictions),
            "R2": r2_score(y_val, predictions),
        })

        model_filename = (
            f"{model_name.lower().replace('-', '_')}_{horizon}.joblib"
        )
        joblib.dump(model, MODELS_DIR / model_filename)

lgbm_results_df = pd.DataFrame(lgbm_results)

print("LIGHTGBM — VALIDATION PERFORMANCE")
print("-" * 75)
print(
    lgbm_results_df
    .sort_values(["Horizon", "Model"])
    .round(6)
    .to_string(index=False)
)

print("\nModels saved in:", MODELS_DIR)
```

```python
# Cell 38: Full-dataset date and duplicate audit using chunks

import pandas as pd
from collections import Counter

news_path = PROJECT_DIR.parent / "processed_news_dataset.csv"

chunk_size = 100_000
total_rows = 0
invalid_dates = 0
missing_values = Counter()
date_counts = Counter()
seen_urls = set()
duplicate_urls = 0
empty_text_rows = 0

for chunk in pd.read_csv(
    news_path,
    chunksize=chunk_size,
    low_memory=False
):
    total_rows += len(chunk)

    for col in ["date", "title", "news", "url"]:
        missing_values[col] += chunk[col].isna().sum()

    parsed_dates = pd.to_datetime(chunk["date"], errors="coerce")
    invalid_dates += parsed_dates.isna().sum()

    date_counts.update(
        parsed_dates.dropna().dt.strftime("%Y-%m-%d").tolist()
    )

    empty_text_rows += (
        chunk["title"].fillna("").str.strip().eq("")
        | chunk["news"].fillna("").str.strip().eq("")
    ).sum()

    urls = chunk["url"].dropna().astype(str)
    for url in urls:
        if url in seen_urls:
            duplicate_urls += 1
        else:
            seen_urls.add(url)

print("FULL NEWS DATASET AUDIT")
print("-" * 55)
print(f"Total rows: {total_rows:,}")
print(f"Invalid dates: {invalid_dates:,}")
print(f"Duplicate URLs: {duplicate_urls:,}")
print(f"Rows with empty title or news: {empty_text_rows:,}")

print("\nMissing values:")
for col, count in missing_values.items():
    print(f"{col}: {count:,}")

if date_counts:
    print("\nDate coverage:")
    print("First date:", min(date_counts))
    print("Last date:", max(date_counts))
    print("Unique dates:", len(date_counts))

    daily_counts = pd.Series(date_counts).sort_index()
    print("\nArticles per day — summary:")
    print(daily_counts.describe().round(2).to_string())

    print("\nTop 10 dates by article count:")
    print(daily_counts.sort_values(ascending=False).head(10).to_string())

print("\nNews audit complete.")
```

```python
# Cell 40: Conservative news-to-market calendar alignment

import pandas as pd
import numpy as np

news_path = PROJECT_DIR.parent / "processed_news_dataset.csv"
market_path = FEATURES_DIR / "market_volatility_features.parquet"

# 1. Load market trading dates
market_df = pd.read_parquet(market_path)
market_df["Date"] = pd.to_datetime(
    market_df["Date"]
).dt.normalize()

market_dates = pd.DatetimeIndex(
    sorted(market_df["Date"].dropna().unique())
)

# 2. Aggregate news articles by their recorded date
news_df = pd.read_csv(
    news_path,
    usecols=["date", "title", "news"],
    parse_dates=["date"],
    low_memory=False
)

news_df["Date"] = pd.to_datetime(
    news_df["date"], errors="coerce"
).dt.normalize()

news_df["combined_text"] = (
    news_df["title"].fillna("").astype(str).str.strip()
    + ". "
    + news_df["news"].fillna("").astype(str).str.strip()
)

daily_news_features = (
    news_df.groupby("Date")
    .agg(
        news_count=("combined_text", "size"),
        mean_article_chars=("combined_text", lambda s: s.str.len().mean()),
    )
    .reset_index()
)

# 3. Shift news availability to the next market trading session
daily_news_features["available_from"] = (
    daily_news_features["Date"].map(
        lambda d: (
            market_dates[market_dates > d][0]
            if (market_dates > d).any()
            else pd.NaT
        )
    )
)

daily_news_features = daily_news_features.dropna(
    subset=["available_from"]
)

daily_news_features["available_from"] = pd.to_datetime(
    daily_news_features["available_from"]
).dt.normalize()

# 4. Aggregate news that becomes available on each trading date
available_news = (
    daily_news_features.groupby("available_from")
    .agg(
        news_count=("news_count", "sum"),
        mean_article_chars=("mean_article_chars", "mean"),
    )
    .reset_index()
    .rename(columns={"available_from": "Date"})
)

# 5. Merge onto all stock-level market observations
market_news = market_df.merge(
    available_news,
    on="Date",
    how="left",
    validate="many_to_one"
)

market_news["news_count"] = (
    market_news["news_count"].fillna(0).astype(int)
)

# Keep missing article-length values as NaN when there is no news
market_news["has_news"] = (
    market_news["news_count"] > 0
).astype(int)

# 6. Save aligned dataset
output_csv = FEATURES_DIR / "market_daily_news_aligned.csv"
output_parquet = FEATURES_DIR / "market_daily_news_aligned.parquet"

market_news.to_csv(output_csv, index=False)
market_news.to_parquet(output_parquet, index=False)

print("NEWS-MARKET ALIGNMENT")
print("-" * 55)
print(f"Market trading dates: {len(market_dates):,}")
print(f"News dates: {daily_news_features['Date'].nunique():,}")
print(f"News dates aligned: {available_news['Date'].nunique():,}")
print(f"Market rows: {len(market_news):,}")
print(f"Rows with available news: {(market_news['news_count'] > 0).sum():,}")
print(f"Saved: {output_parquet}")
```

```python
# Cell 42: Create a date-stratified sample for LLM feature extraction

import pandas as pd
import numpy as np

news_path = PROJECT_DIR.parent / "processed_news_dataset.csv"

# Read only the columns needed for feature extraction
news_df = pd.read_csv(
    news_path,
    usecols=["date", "title", "news", "url"],
    parse_dates=["date"],
    low_memory=False
)

news_df["date"] = pd.to_datetime(
    news_df["date"], errors="coerce"
).dt.normalize()

news_df = news_df.dropna(
    subset=["date", "title", "news", "url"]
).copy()

# Combine title and body while preserving both original fields
news_df["combined_text"] = (
    news_df["title"].astype(str).str.strip()
    + ". "
    + news_df["news"].astype(str).str.strip()
)

# Stratify by month so the sample covers the complete period
news_df["month"] = news_df["date"].dt.to_period("M").astype(str)

sample_size = min(1000, len(news_df))
rng = np.random.default_rng(42)

monthly_groups = list(news_df.groupby("month", sort=True))
monthly_sizes = np.array([len(group) for _, group in monthly_groups])

# Allocate sample slots proportionally, with at least one per month
allocations = np.floor(
    sample_size * monthly_sizes / monthly_sizes.sum()
).astype(int)

allocations = np.maximum(allocations, 1)

# Adjust allocations to equal the requested sample size
while allocations.sum() > sample_size:
    candidates = np.where(allocations > 1)[0]
    if len(candidates) == 0:
        break
    idx = candidates[np.argmax(allocations[candidates])]
    allocations[idx] -= 1

while allocations.sum() < sample_size:
    capacities = monthly_sizes - allocations
    candidates = np.where(capacities > 0)[0]
    if len(candidates) == 0:
        break
    idx = candidates[np.argmax(capacities[candidates])]
    allocations[idx] += 1

sample_parts = []

for (month, group), n in zip(monthly_groups, allocations):
    n = min(int(n), len(group))
    sample_parts.append(
        group.sample(n=n, random_state=42)
    )

llm_sample = (
    pd.concat(sample_parts, ignore_index=True)
    .sort_values(["date", "url"])
    .reset_index(drop=True)
)

# Assign stable IDs for auditing and later joins
llm_sample.insert(
    0, "article_id",
    [f"NEWS_{i:05d}" for i in range(1, len(llm_sample) + 1)]
)

# Save the sample
LLM_SAMPLE_PATH = FEATURES_DIR / "llm_news_sample_1000.csv"
llm_sample.to_csv(LLM_SAMPLE_PATH, index=False)

print("LLM SAMPLE PREPARATION")
print("-" * 50)
print(f"Original articles: {len(news_df):,}")
print(f"Sampled articles: {len(llm_sample):,}")
print(f"Unique dates: {llm_sample['date'].nunique():,}")
print(f"Date range: {llm_sample['date'].min().date()} "
      f"to {llm_sample['date'].max().date()}")
print(f"Duplicate URLs in sample: {llm_sample['url'].duplicated().sum()}")
print(f"Saved to: {LLM_SAMPLE_PATH}")

display(
    llm_sample[["article_id", "date", "title", "url"]].head(10)
)
```

```python
# Cell 53: Validated LLM extraction function

import requests
import json
import pandas as pd
import time

OLLAMA_URL = "http://localhost:11434/api/generate"
MODEL_NAME = "qwen2.5:1.5b"

VALID_SENTIMENTS = {"negative", "neutral", "positive"}

VALID_CATEGORIES = {
    "earnings", "merger_acquisition", "regulation_policy",
    "macroeconomics", "interest_rates", "inflation",
    "commodity_energy", "market_movement", "analyst_rating",
    "corporate_action", "litigation", "geopolitical",
    "technology", "other"
}

VALID_SCOPES = {
    "individual_company", "sector", "market_wide", "unclear"
}

def extract_news_features(title, article_text, retries=2):
    """Extract structured financial-news features and validate them."""

    prompt = f"""
You are a financial-news analyst studying Indian equity-market volatility.

TITLE: {str(title)[:500]}
ARTICLE: {str(article_text)[:2500]}

Classify only using evidence in the supplied text.
Do not invent facts. Distinguish market relevance from sentiment.

Return JSON with exactly these fields:
market_relevance: integer 0-2
sentiment: negative, neutral, or positive
uncertainty: integer 0-4
materiality: integer 0-4
event_category: earnings, merger_acquisition, regulation_policy,
macroeconomics, interest_rates, inflation, commodity_energy,
market_movement, analyst_rating, corporate_action, litigation,
geopolitical, technology, or other
affected_scope: individual_company, sector, market_wide, or unclear
reason: short evidence-based explanation

Scoring:
market_relevance: 0 irrelevant, 1 indirect, 2 direct.
uncertainty: 0 very low, 4 very high.
materiality: 0 negligible, 4 potentially very high.
Return valid JSON only.
"""

    last_error = None

    for attempt in range(retries + 1):
        try:
            response = requests.post(
                OLLAMA_URL,
                json={
                    "model": MODEL_NAME,
                    "prompt": prompt,
                    "stream": False,
                    "format": "json",
                    "options": {
                        "temperature": 0,
                        "num_predict": 220
                    }
                },
                timeout=180
            )
            response.raise_for_status()
            output = json.loads(response.json()["response"])

            required = {
                "market_relevance", "sentiment", "uncertainty",
                "materiality", "event_category", "affected_scope",
                "reason"
            }

            if not isinstance(output, dict) or not required.issubset(output):
                raise ValueError("Missing required JSON fields")

            # Validate integer scores (booleans are not accepted)
            limits = {
                "market_relevance": 2,
                "uncertainty": 4,
                "materiality": 4
            }

            for field, upper in limits.items():
                value = output[field]
                if (
                    isinstance(value, bool)
                    or not isinstance(value, int)
                    or not 0 <= value <= upper
                ):
                    raise ValueError(f"Invalid score: {field}={value}")

            if output["sentiment"] not in VALID_SENTIMENTS:
                raise ValueError("Invalid sentiment")

            if output["event_category"] not in VALID_CATEGORIES:
                raise ValueError("Invalid event category")

            if output["affected_scope"] not in VALID_SCOPES:
                raise ValueError("Invalid affected scope")

            if not isinstance(output["reason"], str):
                raise ValueError("Reason must be text")

            return output

        except Exception as error:
            last_error = str(error)
            if attempt < retries:
                time.sleep(2)

    raise RuntimeError(
        f"Extraction failed after {retries + 1} attempts: {last_error}"
    )

print("Validated extraction function is ready.")
print("It checks JSON fields, score ranges, and allowed labels.")
```

```python
# Cell 54: Safe, resumable 10-article LLM pilot

import pandas as pd
import json
import csv

OUTPUT_PATH = FEATURES_DIR / "llm_news_features_results.csv"
CHECKPOINT_PATH = FEATURES_DIR / "llm_extraction_checkpoint.json"
SAMPLE_PATH = FEATURES_DIR / "llm_news_sample_1000.csv"

news_sample = pd.read_csv(SAMPLE_PATH, dtype={"article_id": str})
news_sample["article_id"] = news_sample["article_id"].astype(str)

expected_columns = [
    "article_id", "market_relevance", "sentiment",
    "uncertainty", "materiality", "event_category",
    "affected_scope", "reason", "extraction_status"
]

# Load existing results safely if the file is readable.
if OUTPUT_PATH.exists():
    try:
        existing = pd.read_csv(
            OUTPUT_PATH,
            dtype={"article_id": str},
            engine="python",
            on_bad_lines="skip"
        )
        existing = existing.reindex(columns=expected_columns)
    except Exception:
        existing = pd.DataFrame(columns=expected_columns)
else:
    existing = pd.DataFrame(columns=expected_columns)

# Retain successful records; retry failed or absent records.
existing["article_id"] = existing["article_id"].astype("string")
successful_ids = set(
    existing.loc[
        existing["extraction_status"].eq("success"),
        "article_id"
    ].dropna().astype(str)
)

pending = news_sample[
    ~news_sample["article_id"].isin(successful_ids)
].head(10)

print("Articles selected:", len(pending))
print("Previously successful:", len(successful_ids))

for _, row in pending.iterrows():
    article_id = str(row["article_id"])
    title = str(row.get("title", ""))
    article_text = str(row.get("news", ""))

    try:
        features = extract_news_features(title, article_text)
        record = {
            "article_id": article_id,
            **features,
            "extraction_status": "success"
        }
        print(f"{article_id}: success")

    except Exception as error:
        record = {
            "article_id": article_id,
            "market_relevance": None,
            "sentiment": None,
            "uncertainty": None,
            "materiality": None,
            "event_category": None,
            "affected_scope": None,
            "reason": str(error),
            "extraction_status": "failed"
        }
        print(f"{article_id}: failed — {error}")

    # Update in memory, replacing any older record for this article.
    existing = existing[
        existing["article_id"].astype(str) != article_id
    ]
    existing = pd.concat(
        [existing, pd.DataFrame([record])],
        ignore_index=True
    )

    # Write a complete, correctly quoted CSV after every article.
    existing.reindex(columns=expected_columns).to_csv(
        OUTPUT_PATH,
        index=False,
        quoting=csv.QUOTE_ALL,
        encoding="utf-8"
    )

successful_ids = set(
    existing.loc[
        existing["extraction_status"].eq("success"),
        "article_id"
    ].dropna().astype(str)
)

checkpoint = {
    "model": MODEL_NAME,
    "total_articles": int(len(news_sample)),
    "completed_articles": len(successful_ids),
    "remaining_articles": int(len(news_sample) - len(successful_ids)),
    "output_file": str(OUTPUT_PATH)
}

CHECKPOINT_PATH.write_text(
    json.dumps(checkpoint, indent=2),
    encoding="utf-8"
)

print("\nPilot run finished.")
print("Successful articles:", len(successful_ids))
print("Results saved:", OUTPUT_PATH)
```

```python
# Cell 63: Construct intraday + overnight variance proxy

import pandas as pd
import numpy as np

MARKET_PATH = FEATURES_DIR / "market_volatility_features.parquet"
OUTPUT_PATH = FEATURES_DIR / "market_combined_volatility_features.parquet"

market = pd.read_parquet(MARKET_PATH).copy()
market["Date"] = pd.to_datetime(market["Date"])
market = market.sort_values(["Symbol", "Date"]).copy()

# Previous raw close within each stock
market["previous_close_raw"] = (
    market.groupby("Symbol")["Close"].shift(1)
)

# Overnight log return: ln(Open_t / Close_{t-1})
market["overnight_log_return"] = np.log(
    market["Open"] / market["previous_close_raw"]
)

# Overnight variance proxy
market["overnight_variance"] = (
    market["overnight_log_return"] ** 2
)

# Combined intraday + overnight variance proxy
market["combined_variance"] = (
    market["gk_variance"] + market["overnight_variance"]
)

# Validate non-negative variance and price availability
valid = (
    market["Open"].gt(0)
    & market["Close"].gt(0)
    & market["gk_variance"].notna()
    & market["combined_variance"].notna()
)

print("Total rows:", len(market))
print("Rows with valid combined variance:", int(valid.sum()))
print("Missing combined variance:", int(market["combined_variance"].isna().sum()))
print("Negative combined variance:", int((market["combined_variance"] < 0).sum()))

print("\nVariance component summary:")
display(
    market[
        ["gk_variance", "overnight_variance", "combined_variance"]
    ].describe(percentiles=[0.50, 0.90, 0.99]).T
)

market.to_parquet(OUTPUT_PATH, index=False)

print("\nSaved combined volatility features to:")
print(OUTPUT_PATH)
```

```python
# Cell 65: Create future combined-variance targets by stock

import pandas as pd
import numpy as np

INPUT_PATH = FEATURES_DIR / "market_combined_volatility_features.parquet"
OUTPUT_PATH = FEATURES_DIR / "market_combined_volatility_targets.parquet"

market = pd.read_parquet(INPUT_PATH).copy()
market["Date"] = pd.to_datetime(market["Date"])
market = market.sort_values(["Symbol", "Date"]).copy()

grouped = market.groupby("Symbol", sort=False)["combined_variance"]

# Next trading day's combined variance
market["target_combined_var_next1d"] = grouped.shift(-1)

# Mean combined variance over the next five trading days.
# Shift first so day t is excluded from its own future target.
future_values = pd.concat(
    [
        grouped.shift(-h).rename(f"h{h}")
        for h in range(1, 6)
    ],
    axis=1
)

market["target_combined_var_next5d"] = future_values.mean(
    axis=1,
    skipna=False
)

# Save the new targets without overwriting the original dataset
market.to_parquet(OUTPUT_PATH, index=False)

print("Rows:", len(market))
print(
    "Valid next-day targets:",
    int(market["target_combined_var_next1d"].notna().sum())
)
print(
    "Valid next-5-day targets:",
    int(market["target_combined_var_next5d"].notna().sum())
)
print(
    "Missing next-day targets:",
    int(market["target_combined_var_next1d"].isna().sum())
)
print(
    "Missing next-5-day targets:",
    int(market["target_combined_var_next5d"].isna().sum())
)
print("\nTarget preview:")
display(
    market[
        [
            "Date",
            "Symbol",
            "gk_variance",
            "overnight_variance",
            "combined_variance",
            "target_combined_var_next1d",
            "target_combined_var_next5d"
        ]
    ].head(10)
)

print("\nSaved to:", OUTPUT_PATH)
```

```python
# Cell 67: Prepare combined-variance dataset and chronological splits

import pandas as pd
import numpy as np

market = pd.read_parquet(
    FEATURES_DIR / "market_combined_volatility_targets.parquet"
).copy()

market["Date"] = pd.to_datetime(market["Date"])
market = market.sort_values(["Date", "Symbol"]).reset_index(drop=True)

# Historical predictors available at time t
market["combined_var_lag1"] = (
    market.groupby("Symbol")["combined_variance"].shift(1)
)

market["combined_var_mean5"] = (
    market.groupby("Symbol")["combined_variance"]
    .transform(lambda s: s.shift(1).rolling(5, min_periods=5).mean())
)

market["combined_var_mean22"] = (
    market.groupby("Symbol")["combined_variance"]
    .transform(lambda s: s.shift(1).rolling(22, min_periods=22).mean())
)

# Retain rows with all predictors and both targets available
model_data = market.dropna(
    subset=[
        "combined_var_lag1",
        "combined_var_mean5",
        "combined_var_mean22",
        "target_combined_var_next1d",
        "target_combined_var_next5d"
    ]
).copy()

# Split chronologically by unique dates, not random rows
unique_dates = np.sort(model_data["Date"].unique())
n_dates = len(unique_dates)

train_cut = unique_dates[int(n_dates * 0.70)]
val_cut = unique_dates[int(n_dates * 0.85)]

# Five-observation purge before validation and test windows
purge_days = 5
val_start_idx = np.searchsorted(unique_dates, train_cut, side="right") + purge_days
test_start_idx = np.searchsorted(unique_dates, val_cut, side="right") + purge_days

val_start_date = unique_dates[min(val_start_idx, n_dates - 1)]
test_start_date = unique_dates[min(test_start_idx, n_dates - 1)]

train = model_data[model_data["Date"] <= train_cut].copy()
validation = model_data[
    (model_data["Date"] > train_cut) &
    (model_data["Date"] < val_start_date)
].copy()
test = model_data[model_data["Date"] >= test_start_date].copy()

# Print partition details
print("Model dataset shape:", model_data.shape)
print("Unique dates:", n_dates)
print("\nTrain:", train.shape, "|", train["Date"].min(), "to", train["Date"].max())
print("Validation:", validation.shape, "|", validation["Date"].min(), "to", validation["Date"].max())
print("Test:", test.shape, "|", test["Date"].min(), "to", test["Date"].max())

print("\nFeature columns:")
print(["combined_var_lag1", "combined_var_mean5", "combined_var_mean22"])
```

```python
# Cell 68: Correct chronological split with five-date purge gaps

import numpy as np
import pandas as pd

# Reuse model_data created in Cell 67
unique_dates = np.sort(model_data["Date"].unique())
n_dates = len(unique_dates)

# Candidate chronological boundaries: 70% train, 85% train+validation
train_end_idx = int(n_dates * 0.70)
val_end_idx = int(n_dates * 0.85)

train_end_date = unique_dates[train_end_idx - 1]
val_end_date = unique_dates[val_end_idx - 1]

# Five trading-date observations excluded at each boundary
train = model_data[
    model_data["Date"] <= train_end_date
].copy()

validation = model_data[
    (model_data["Date"] > unique_dates[train_end_idx + 4]) &
    (model_data["Date"] <= val_end_date)
].copy()

test = model_data[
    model_data["Date"] > unique_dates[val_end_idx + 4]
].copy()

print("Total model rows:", len(model_data))
print("Total unique dates:", n_dates)

print("\nTRAIN")
print("Rows:", len(train))
print("Dates:", train["Date"].min(), "to", train["Date"].max())
print("Unique dates:", train["Date"].nunique())

print("\nVALIDATION")
print("Rows:", len(validation))
print("Dates:", validation["Date"].min(), "to", validation["Date"].max())
print("Unique dates:", validation["Date"].nunique())

print("\nTEST")
print("Rows:", len(test))
print("Dates:", test["Date"].min(), "to", test["Date"].max())
print("Unique dates:", test["Date"].nunique())

print("\nBoundary checks:")
print("Train ends before validation:", train["Date"].max() < validation["Date"].min())
print("Validation ends before test:", validation["Date"].max() < test["Date"].min())
```

```python
# Cell 69: HAR-style linear regression for combined variance

import numpy as np
import pandas as pd
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score

FEATURES = [
    "combined_var_lag1",
    "combined_var_mean5",
    "combined_var_mean22"
]

TARGETS = {
    "Next 1 day": "target_combined_var_next1d",
    "Next 5 days": "target_combined_var_next5d"
}

results = []
fitted_models = {}

for horizon, target in TARGETS.items():
    model = LinearRegression()

    X_train = train[FEATURES]
    y_train = train[target]

    X_val = validation[FEATURES]
    y_val = validation[target]

    model.fit(X_train, y_train)
    predictions = model.predict(X_val)

    fitted_models[horizon] = model

    results.append({
        "Horizon": horizon,
        "RMSE": np.sqrt(mean_squared_error(y_val, predictions)),
        "MAE": mean_absolute_error(y_val, predictions),
        "R2": r2_score(y_val, predictions)
    })

    print(f"\n{horizon}")
    print("Intercept:", model.intercept_)
    print("Coefficients:")
    for feature, coefficient in zip(FEATURES, model.coef_):
        print(f"  {feature}: {coefficient:.6f}")

results_df = pd.DataFrame(results)
print("\nValidation performance:")
display(results_df)
```

```python
# Cell 70: Compare HAR against a training-mean baseline

import numpy as np
import pandas as pd
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score

comparison = []

for horizon, target in TARGETS.items():
    y_train = train[target]
    y_val = validation[target]

    # Baseline: predict training-set mean
    mean_prediction = np.full(len(y_val), y_train.mean())

    # HAR predictions from Cell 69
    har_prediction = fitted_models[horizon].predict(validation[FEATURES])

    for model_name, prediction in [
        ("Training-mean baseline", mean_prediction),
        ("HAR linear regression", har_prediction)
    ]:
        comparison.append({
            "Horizon": horizon,
            "Model": model_name,
            "RMSE": np.sqrt(mean_squared_error(y_val, prediction)),
            "MAE": mean_absolute_error(y_val, prediction),
            "R2": r2_score(y_val, prediction)
        })

comparison_df = pd.DataFrame(comparison)

display(
    comparison_df.sort_values(["Horizon", "RMSE"]).reset_index(drop=True)
)

print("\nHAR RMSE improvement relative to baseline:")
for horizon in TARGETS:
    subset = comparison_df[comparison_df["Horizon"] == horizon].set_index("Model")
    baseline_rmse = subset.loc["Training-mean baseline", "RMSE"]
    har_rmse = subset.loc["HAR linear regression", "RMSE"]

    improvement = 100 * (baseline_rmse - har_rmse) / baseline_rmse
    print(f"{horizon}: {improvement:.2f}%")
```

```python
# Cell 72: HAR regression with logarithmically transformed targets

import numpy as np
import pandas as pd
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score

FEATURES = [
    "combined_var_lag1",
    "combined_var_mean5",
    "combined_var_mean22"
]

# Small positive constant to make log(variance + epsilon) defined
EPSILON = 1e-10

log_models = {}
log_results = []

for horizon, target in TARGETS.items():
    model = LinearRegression()

    X_train = train[FEATURES]
    X_val = validation[FEATURES]

    y_train_log = np.log(train[target] + EPSILON)

    model.fit(X_train, y_train_log)

    # Convert log-scale predictions back to the original variance scale
    pred_log = model.predict(X_val)
    pred_variance = np.maximum(np.exp(pred_log) - EPSILON, 0)

    y_val = validation[target]

    log_models[horizon] = model

    log_results.append({
        "Horizon": horizon,
        "RMSE": np.sqrt(mean_squared_error(y_val, pred_variance)),
        "MAE": mean_absolute_error(y_val, pred_variance),
        "R2": r2_score(y_val, pred_variance)
    })

    print(f"\n{horizon}")
    print("Log-scale intercept:", model.intercept_)
    print("Log-scale coefficients:")
    for feature, coef in zip(FEATURES, model.coef_):
        print(f"  {feature}: {coef:.6f}")

log_results_df = pd.DataFrame(log_results)

print("\nLog-HAR validation performance (original variance scale):")
display(log_results_df)

print("\nComparison with ordinary HAR:")
display(
    comparison_df[["Horizon", "Model", "RMSE", "MAE", "R2"]]
    .merge(
        log_results_df,
        on="Horizon",
        suffixes=("_previous", "_logHAR")
    )
)
```

```python
# Cell 74: Diagnose log-HAR forecast errors by volatility regime

import numpy as np
import pandas as pd
from sklearn.metrics import mean_squared_error, mean_absolute_error

error_diagnostics = []

for horizon, target in TARGETS.items():
    y_true = validation[target].to_numpy()
    y_pred = np.maximum(
        np.exp(log_models[horizon].predict(validation[FEATURES])) - EPSILON,
        0
    )

    residuals = y_true - y_pred

    # Define high-volatility observations using the validation target's 90th percentile
    threshold = np.quantile(y_true, 0.90)
    regimes = {
        "Lower 90%": y_true <= threshold,
        "Upper 10%": y_true > threshold
    }

    print(f"\n{horizon}")
    print("Mean forecast error (actual - predicted):", residuals.mean())
    print("Median forecast error:", np.median(residuals))
    print("90th-percentile target threshold:", threshold)

    for regime_name, mask in regimes.items():
        error_diagnostics.append({
            "Horizon": horizon,
            "Regime": regime_name,
            "Observations": int(mask.sum()),
            "RMSE": np.sqrt(mean_squared_error(y_true[mask], y_pred[mask])),
            "MAE": mean_absolute_error(y_true[mask], y_pred[mask]),
            "Mean_Error": residuals[mask].mean()
        })

diagnostics_df = pd.DataFrame(error_diagnostics)

print("\nForecast errors by target-volatility regime:")
display(diagnostics_df)
```

```python
# Cell 75: Evaluate log-HAR with Duan's smearing correction

import numpy as np
import pandas as pd
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score

smearing_results = []

for horizon, target in TARGETS.items():
    model = log_models[horizon]

    # Training residuals on the log scale
    train_log_actual = np.log(train[target].to_numpy() + EPSILON)
    train_log_pred = model.predict(train[FEATURES])
    train_log_residuals = train_log_actual - train_log_pred

    # Smearing factor estimated from training data only
    smearing_factor = np.mean(np.exp(train_log_residuals))

    # Validation predictions with retransformation correction
    val_log_pred = model.predict(validation[FEATURES])
    val_pred = np.maximum(
        np.exp(val_log_pred) * smearing_factor - EPSILON,
        0
    )

    y_val = validation[target].to_numpy()

    rmse = np.sqrt(mean_squared_error(y_val, val_pred))
    mae = mean_absolute_error(y_val, val_pred)
    r2 = r2_score(y_val, val_pred)

    smearing_results.append({
        "Horizon": horizon,
        "Smearing_Factor": smearing_factor,
        "RMSE": rmse,
        "MAE": mae,
        "R2": r2
    })

smearing_df = pd.DataFrame(smearing_results)

print("Log-HAR with Duan smearing correction:")
display(smearing_df)

print("\nComparison with uncorrected log-HAR:")
display(
    log_results_df[
        ["Horizon", "RMSE", "MAE", "R2"]
    ].merge(
        smearing_df[
            ["Horizon", "RMSE", "MAE", "R2"]
        ],
        on="Horizon",
        suffixes=("_uncorrected", "_smearing")
    )
)
```

```python
# Cell 79: GARCH residual diagnostics without statsmodels

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy.stats import chi2

standardized_residuals = pd.Series(
    np.asarray(garch_fit.std_resid)
).dropna().reset_index(drop=True)

def ljung_box_manual(series, lags=(10, 20)):
    """
    Ljung-Box Q statistic:
    Q(m) = n(n+2) * sum[r_k^2 / (n-k)]
    """
    x = np.asarray(series, dtype=float)
    x = x[np.isfinite(x)]
    n = len(x)

    # Sample autocorrelations at the required lags
    centered = x - x.mean()
    denominator = np.dot(centered, centered)

    if denominator == 0:
        raise ValueError("Series has zero variance.")

    max_lag = max(lags)
    acf = np.array([
        np.dot(centered[k:], centered[:-k]) / denominator
        for k in range(1, max_lag + 1)
    ])

    results = []

    for m in lags:
        q_stat = n * (n + 2) * np.sum(
            acf[:m] ** 2 / (n - np.arange(1, m + 1))
        )
        p_value = chi2.sf(q_stat, df=m)

        results.append({
            "Lag": m,
            "Ljung_Box_Q": q_stat,
            "Degrees_of_Freedom": m,
            "p_value": p_value
        })

    return pd.DataFrame(results)

# Test serial correlation in standardized residuals
print("Ljung-Box test: standardized residuals")
display(ljung_box_manual(standardized_residuals))

# Test remaining volatility clustering
print("\nLjung-Box test: squared standardized residuals")
display(ljung_box_manual(standardized_residuals ** 2))

# Visual diagnostics
fig, axes = plt.subplots(2, 1, figsize=(11, 7))

axes[0].plot(standardized_residuals)
axes[0].axhline(0, linestyle="--")
axes[0].set_title("GARCH Standardized Residuals")
axes[0].set_ylabel("Residual")
axes[0].grid(alpha=0.25)

axes[1].plot(standardized_residuals ** 2)
axes[1].set_title("Squared Standardized Residuals")
axes[1].set_ylabel("Squared residual")
axes[1].set_xlabel("Observation")
axes[1].grid(alpha=0.25)

plt.tight_layout()

figure_path = FIGURES_DIR / "garch_residual_diagnostics_adanient.png"
plt.savefig(figure_path, dpi=300, bbox_inches="tight")
plt.show()

print("Figure saved to:", figure_path)
```

```python
# Cell 81: Compare GARCH and GJR-GARCH residual diagnostics

import numpy as np
import pandas as pd

def residual_diagnostic_table(fit, model_name):
    z = pd.Series(np.asarray(fit.std_resid)).dropna().reset_index(drop=True)

    residual_tests = ljung_box_manual(z, lags=(10, 20))
    residual_tests["Series"] = "Standardized residuals"

    squared_tests = ljung_box_manual(z**2, lags=(10, 20))
    squared_tests["Series"] = "Squared standardized residuals"

    results = pd.concat(
        [residual_tests, squared_tests],
        ignore_index=True
    )
    results.insert(0, "Model", model_name)
    return results[
        ["Model", "Series", "Lag", "Ljung_Box_Q", "p_value"]
    ]

diagnostics_comparison = pd.concat(
    [
        residual_diagnostic_table(garch_fit, "GARCH(1,1)"),
        residual_diagnostic_table(gjr_fit, "GJR-GARCH(1,1)")
    ],
    ignore_index=True
)

display(diagnostics_comparison)

print("\nInformation criteria:")
display(pd.DataFrame([
    {"Model": "GARCH(1,1)", "AIC": garch_fit.aic, "BIC": garch_fit.bic},
    {"Model": "GJR-GARCH(1,1)", "AIC": gjr_fit.aic, "BIC": gjr_fit.bic}
]))
```

```python
# Cell 84: Expanding-window one-step-ahead forecasts

import numpy as np
import pandas as pd
from arch import arch_model

SYMBOL = "ADANIENT.NS"

stock = market.loc[
    market["Symbol"] == SYMBOL,
    ["Date", "log_return"]
].copy()

stock["Date"] = pd.to_datetime(stock["Date"])
stock = stock.sort_values("Date").dropna(subset=["log_return"])

train_stock = stock[stock["Date"] <= train_end_date].copy()
test_stock = stock[stock["Date"] > test_start_date].copy()

history = list(train_stock["log_return"].to_numpy() * 100)
test_scaled = test_stock["log_return"].to_numpy() * 100

forecast_rows = []

for i, actual_return in enumerate(test_scaled):
    # Fit using only returns observed up to the current forecast origin
    garch_fit_i = arch_model(
        np.asarray(history),
        mean="Constant",
        vol="GARCH",
        p=1,
        q=1,
        dist="t",
        rescale=False
    ).fit(disp="off")

    gjr_fit_i = arch_model(
        np.asarray(history),
        mean="Constant",
        vol="GARCH",
        p=1,
        o=1,
        q=1,
        dist="t",
        rescale=False
    ).fit(disp="off")

    # One-step-ahead conditional variance forecasts
    garch_var = float(
        garch_fit_i.forecast(horizon=1, reindex=False).variance.iloc[-1, 0]
    ) / 10000

    gjr_var = float(
        gjr_fit_i.forecast(horizon=1, reindex=False).variance.iloc[-1, 0]
    ) / 10000

    forecast_rows.append({
        "Date": test_stock.iloc[i]["Date"],
        "Actual_Squared_Return": (actual_return / 100) ** 2,
        "GARCH_Forecast_Variance": garch_var,
        "GJR_GARCH_Forecast_Variance": gjr_var
    })

    # The actual return is added only after that day's forecast is produced
    history.append(actual_return)

garch_oos_results = pd.DataFrame(forecast_rows)

OUTPUT_PATH = TABLES_DIR / "adanient_garch_oos_forecasts.csv"
garch_oos_results.to_csv(OUTPUT_PATH, index=False)

print("Forecasts generated:", len(garch_oos_results))
print("Forecast date range:",
      garch_oos_results["Date"].min(),
      "to",
      garch_oos_results["Date"].max())
print("Saved to:", OUTPUT_PATH)

display(garch_oos_results.head())
```

```python
# Cell 85: Out-of-sample GARCH evaluation

import numpy as np
import pandas as pd
from sklearn.metrics import mean_squared_error, mean_absolute_error

df = garch_oos_results.copy()

actual = df["Actual_Squared_Return"].to_numpy()

forecast_columns = {
    "GARCH(1,1)": "GARCH_Forecast_Variance",
    "GJR-GARCH(1,1)": "GJR_GARCH_Forecast_Variance"
}

evaluation = []

for model_name, forecast_col in forecast_columns.items():
    predicted = np.maximum(
        df[forecast_col].to_numpy(),
        1e-12
    )

    actual_safe = np.maximum(actual, 0)

    # QLIKE, using squared return as the realized-variance proxy
    qlike = np.mean(
        np.log(predicted) + actual_safe / predicted
    )

    evaluation.append({
        "Model": model_name,
        "RMSE": np.sqrt(mean_squared_error(actual, predicted)),
        "MAE": mean_absolute_error(actual, predicted),
        "QLIKE": qlike
    })

garch_evaluation = pd.DataFrame(evaluation)

print("ADANIENT out-of-sample evaluation:")
display(garch_evaluation.round(8))

print("\nBest model by each metric:")
for metric in ["RMSE", "MAE", "QLIKE"]:
    best = garch_evaluation.loc[
        garch_evaluation[metric].idxmin(), "Model"
    ]
    print(f"{metric}: {best}")
```

```python
# Cell 92: Reusable LLM classifier with JSON validation and retries

import requests
import json
import time

OLLAMA_URL = "http://localhost:11434/api/generate"
LLM_MODEL = "qwen2.5:1.5b"

VALID_SENTIMENTS = {"negative", "neutral", "positive"}

VALID_CATEGORIES = {
    "earnings", "merger_acquisition", "regulation_policy",
    "macroeconomics", "interest_rates", "inflation",
    "commodity_energy", "market_movement", "analyst_rating",
    "corporate_action", "litigation", "geopolitical",
    "technology", "other"
}

VALID_SCOPES = {
    "individual_company", "sector", "market_wide", "unclear"
}

REQUIRED_FIELDS = [
    "market_relevance", "sentiment", "uncertainty",
    "materiality", "event_category", "affected_scope", "reason"
]

def validate_llm_result(result):
    """Validate required fields, types, and permitted categorical values."""

    if not isinstance(result, dict):
        raise ValueError("Output must be a JSON object.")

    missing = [field for field in REQUIRED_FIELDS if field not in result]
    if missing:
        raise ValueError(f"Missing required fields: {missing}")

    integer_ranges = {
        "market_relevance": range(3),
        "uncertainty": range(5),
        "materiality": range(5)
    }

    for field, allowed in integer_ranges.items():
        value = result[field]
        if isinstance(value, bool) or not isinstance(value, int):
            raise ValueError(f"{field} must be an integer.")
        if value not in allowed:
            raise ValueError(f"{field} has an invalid score: {value}")

    if result["sentiment"] not in VALID_SENTIMENTS:
        raise ValueError("Invalid sentiment.")

    if result["event_category"] not in VALID_CATEGORIES:
        raise ValueError("Invalid event_category.")

    if result["affected_scope"] not in VALID_SCOPES:
        raise ValueError("Invalid affected_scope.")

    if not isinstance(result["reason"], str) or not result["reason"].strip():
        raise ValueError("reason must be a non-empty string.")

    return result


def classify_news(article_title, article_text, max_attempts=3):
    """Classify one article, retrying malformed responses or request failures."""

    prompt = f"""
You are an analyst classifying news for Indian equity-market volatility.
Use only evidence in the article. Do not assume every financial article
has maximum relevance or materiality.

SCORING:
market_relevance:
0 = no meaningful connection to Indian equities or markets
1 = indirect, routine, or limited connection
2 = explicit and meaningful information about companies, sectors, or markets

uncertainty: integer 0-4, from none to very high uncertainty.
materiality: integer 0-4, from negligible to exceptionally high impact.
Use high materiality only when the article supports it.

sentiment: negative, neutral, or positive.
event_category: one of {sorted(VALID_CATEGORIES)}.
affected_scope: one of {sorted(VALID_SCOPES)}.

Choose sector for news about a particular industry unless evidence supports
a broader market-wide effect. Choose individual_company for a company-specific
event. Do not infer market-wide scope merely because stock prices are mentioned.

Return only a JSON object with exactly these fields:
{", ".join(REQUIRED_FIELDS)}

ARTICLE TITLE:
{article_title}

ARTICLE TEXT:
{str(article_text)[:3500]}
"""

    last_error = None

    for attempt in range(1, max_attempts + 1):
        try:
            response = requests.post(
                OLLAMA_URL,
                json={
                    "model": LLM_MODEL,
                    "prompt": prompt,
                    "stream": False,
                    "format": "json",
                    "options": {"temperature": 0}
                },
                timeout=180
            )
            response.raise_for_status()

            output = response.json()["response"]
            result = validate_llm_result(json.loads(output))

            return {
                **result,
                "extraction_status": "success",
                "attempts": attempt,
                "error": None
            }

        except Exception as exc:
            last_error = str(exc)
            if attempt < max_attempts:
                time.sleep(1)

    return {
        **{field: None for field in REQUIRED_FIELDS},
        "extraction_status": "failed",
        "attempts": max_attempts,
        "error": last_error
    }


print("Classifier function defined successfully.")
print("Model:", LLM_MODEL)
print("Validation fields:", REQUIRED_FIELDS)
print("No articles processed yet.")
```

```python
# Cell 103: Create historical features and align them with LLM news

import pandas as pd
import numpy as np

market_path = FEATURES_DIR / "market_combined_volatility_targets.parquet"
news_path = FEATURES_DIR / "llm_daily_features_sample.csv"

market_df = pd.read_parquet(market_path)
news_daily = pd.read_csv(news_path, parse_dates=["date"])

market_df["Date"] = pd.to_datetime(market_df["Date"]).dt.normalize()
news_daily["Date"] = pd.to_datetime(news_daily["date"]).dt.normalize()
news_daily = news_daily.drop(columns=["date"])

# Ensure correct chronological order within each stock
market_df = market_df.sort_values(["Symbol", "Date"]).copy()

# Build historical-only predictors: no current/future variance leakage
market_df["combined_var_lag1"] = (
    market_df.groupby("Symbol")["combined_variance"].shift(1)
)

market_df["combined_var_mean5"] = (
    market_df.groupby("Symbol")["combined_variance"]
    .transform(lambda s: s.shift(1).rolling(5, min_periods=5).mean())
)

market_df["combined_var_mean22"] = (
    market_df.groupby("Symbol")["combined_variance"]
    .transform(lambda s: s.shift(1).rolling(22, min_periods=22).mean())
)

# Merge one row of daily news features per date
aligned = market_df.merge(
    news_daily,
    on="Date",
    how="left",
    validate="many_to_one",
    indicator=True
)

aligned["has_sampled_news"] = aligned["_merge"].eq("both").astype(int)
aligned = aligned.drop(columns="_merge")

TARGETS = [
    "target_combined_var_next1d",
    "target_combined_var_next5d"
]

HISTORICAL_FEATURES = [
    "combined_var_lag1",
    "combined_var_mean5",
    "combined_var_mean22"
]

NEWS_FEATURES = [
    "sampled_news_count",
    "mean_market_relevance",
    "mean_sentiment_score",
    "positive_news_count",
    "negative_news_count",
    "mean_uncertainty",
    "mean_materiality",
    "market_movement_count",
    "company_scope_count",
    "sector_scope_count",
    "market_wide_count"
]

# Initial diagnostic: covered dates only, with complete model inputs
common = aligned.loc[aligned["has_sampled_news"].eq(1)].copy()
common = common.replace([np.inf, -np.inf], np.nan)
common = common.dropna(
    subset=HISTORICAL_FEATURES + NEWS_FEATURES + TARGETS
)

common_path = FEATURES_DIR / "market_llm_common_date_diagnostic.parquet"
common.to_parquet(common_path, index=False)

print("Aligned market rows:", len(aligned))
print("Rows with sampled news:", int(aligned["has_sampled_news"].sum()))
print("Complete common-date rows:", len(common))
print("Unique dates retained:", common["Date"].nunique())
print("Unique symbols retained:", common["Symbol"].nunique())
print("Date range:", common["Date"].min(), "to", common["Date"].max())
print("Saved:", common_path)
```

```python
# Cell 105: Historical-only vs LLM-augmented next-day volatility

import numpy as np
import pandas as pd

from sklearn.pipeline import make_pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import Ridge
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score

# Load the fixed chronological splits
train_common = pd.read_parquet(FEATURES_DIR / "llm_common_train.parquet")
validation_common = pd.read_parquet(
    FEATURES_DIR / "llm_common_validation.parquet"
)

HISTORICAL_FEATURES = [
    "combined_var_lag1",
    "combined_var_mean5",
    "combined_var_mean22"
]

NEWS_FEATURES = [
    "sampled_news_count",
    "mean_market_relevance",
    "mean_sentiment_score",
    "positive_news_count",
    "negative_news_count",
    "mean_uncertainty",
    "mean_materiality",
    "market_movement_count",
    "company_scope_count",
    "sector_scope_count",
    "market_wide_count"
]

TARGET = "target_combined_var_next1d"

def prepare_xy(data, features, target):
    clean = data.replace([np.inf, -np.inf], np.nan)
    clean = clean.dropna(subset=[target]).copy()
    return clean[features], clean[target]

X_train_hist, y_train = prepare_xy(
    train_common, HISTORICAL_FEATURES, TARGET
)
X_val_hist, y_val = prepare_xy(
    validation_common, HISTORICAL_FEATURES, TARGET
)

X_train_llm, _ = prepare_xy(
    train_common, HISTORICAL_FEATURES + NEWS_FEATURES, TARGET
)
X_val_llm, _ = prepare_xy(
    validation_common, HISTORICAL_FEATURES + NEWS_FEATURES, TARGET
)

# Both models use the same Ridge regression estimator.
# Imputation and scaling are fitted on training data only.
historical_model = make_pipeline(
    SimpleImputer(strategy="median"),
    StandardScaler(),
    Ridge(alpha=1.0)
)

llm_model = make_pipeline(
    SimpleImputer(strategy="median"),
    StandardScaler(),
    Ridge(alpha=1.0)
)

historical_model.fit(X_train_hist, y_train)
llm_model.fit(X_train_llm, y_train)

pred_hist = historical_model.predict(X_val_hist)
pred_llm = llm_model.predict(X_val_llm)

def evaluate(y_true, pred):
    return {
        "RMSE": np.sqrt(mean_squared_error(y_true, pred)),
        "MAE": mean_absolute_error(y_true, pred),
        "R2": r2_score(y_true, pred)
    }

results = pd.DataFrame([
    {"Model": "Historical-only Ridge", **evaluate(y_val, pred_hist)},
    {"Model": "Historical + LLM Ridge", **evaluate(y_val, pred_llm)}
])

print("Training rows:", len(y_train))
print("Validation rows:", len(y_val))
display(results)

rmse_hist = results.loc[
    results["Model"] == "Historical-only Ridge", "RMSE"
].iloc[0]
rmse_llm = results.loc[
    results["Model"] == "Historical + LLM Ridge", "RMSE"
].iloc[0]

print(
    "Validation RMSE change (%; positive means improvement):",
    round(100 * (rmse_hist - rmse_llm) / rmse_hist, 2)
)
```

```python
# Cell 107: Compare historical, all-LLM, and focused-LLM models

import numpy as np
import pandas as pd

from sklearn.pipeline import make_pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import Ridge
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score

train_common = pd.read_parquet(FEATURES_DIR / "llm_common_train.parquet")
validation_common = pd.read_parquet(
    FEATURES_DIR / "llm_common_validation.parquet"
)

TARGET = "target_combined_var_next1d"

HISTORICAL_FEATURES = [
    "combined_var_lag1",
    "combined_var_mean5",
    "combined_var_mean22"
]

ALL_NEWS_FEATURES = [
    "sampled_news_count",
    "mean_market_relevance",
    "mean_sentiment_score",
    "positive_news_count",
    "negative_news_count",
    "mean_uncertainty",
    "mean_materiality",
    "market_movement_count",
    "company_scope_count",
    "sector_scope_count",
    "market_wide_count"
]

FOCUSED_NEWS_FEATURES = [
    "mean_uncertainty",
    "mean_sentiment_score",
    "negative_news_count"
]

def get_xy(data, features):
    data = data.replace([np.inf, -np.inf], np.nan)
    data = data.dropna(subset=[TARGET]).copy()
    return data[features], data[TARGET]

def fit_evaluate(train_df, val_df, features, model_name):
    X_train, y_train = get_xy(train_df, features)
    X_val, y_val = get_xy(val_df, features)

    model = make_pipeline(
        SimpleImputer(strategy="median"),
        StandardScaler(),
        Ridge(alpha=1.0)
    )

    model.fit(X_train, y_train)
    pred = model.predict(X_val)

    return {
        "Model": model_name,
        "Validation rows": len(y_val),
        "RMSE": np.sqrt(mean_squared_error(y_val, pred)),
        "MAE": mean_absolute_error(y_val, pred),
        "R2": r2_score(y_val, pred)
    }

results = pd.DataFrame([
    fit_evaluate(
        train_common, validation_common,
        HISTORICAL_FEATURES,
        "Historical-only"
    ),
    fit_evaluate(
        train_common, validation_common,
        HISTORICAL_FEATURES + ALL_NEWS_FEATURES,
        "Historical + all LLM features"
    ),
    fit_evaluate(
        train_common, validation_common,
        HISTORICAL_FEATURES + FOCUSED_NEWS_FEATURES,
        "Historical + focused LLM features"
    )
])

baseline_rmse = results.loc[
    results["Model"] == "Historical-only", "RMSE"
].iloc[0]

results["RMSE improvement vs baseline (%)"] = (
    100 * (baseline_rmse - results["RMSE"]) / baseline_rmse
)

display(results.round(6))

print(
    "\nPositive RMSE improvement means lower validation error. "
    "This is exploratory model selection, not final test performance."
)
```

```python
# Cell 111: Date-cluster bootstrap for test RMSE differences

import numpy as np
import pandas as pd

from sklearn.pipeline import make_pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import Ridge

train_common = pd.read_parquet(FEATURES_DIR / "llm_common_train.parquet")
validation_common = pd.read_parquet(
    FEATURES_DIR / "llm_common_validation.parquet"
)
test_common = pd.read_parquet(FEATURES_DIR / "llm_common_test.parquet")

development = pd.concat(
    [train_common, validation_common],
    ignore_index=True
)

HISTORICAL = [
    "combined_var_lag1",
    "combined_var_mean5",
    "combined_var_mean22"
]

ALL_NEWS = [
    "sampled_news_count", "mean_market_relevance",
    "mean_sentiment_score", "positive_news_count",
    "negative_news_count", "mean_uncertainty",
    "mean_materiality", "market_movement_count",
    "company_scope_count", "sector_scope_count", "market_wide_count"
]

FOCUSED_NEWS = [
    "mean_uncertainty", "mean_sentiment_score", "negative_news_count"
]

specs = {
    "Historical-only": HISTORICAL,
    "Historical + all LLM": HISTORICAL + ALL_NEWS,
    "Historical + focused LLM": HISTORICAL + FOCUSED_NEWS
}

def fit_test_predictions(target):
    dev = development.replace([np.inf, -np.inf], np.nan)
    tst = test_common.replace([np.inf, -np.inf], np.nan)

    dev = dev.dropna(subset=[target]).copy()
    tst = tst.dropna(subset=[target]).copy()

    predictions = {}
    y_true = tst[target].to_numpy()

    for name, features in specs.items():
        model = make_pipeline(
            SimpleImputer(strategy="median"),
            StandardScaler(),
            Ridge(alpha=1.0)
        )
        model.fit(dev[features], dev[target])
        predictions[name] = model.predict(tst[features])

    return tst["Date"].to_numpy(), y_true, predictions

def date_cluster_bootstrap(dates, y_true, pred_base, pred_alt,
                           n_boot=2000, seed=42):
    rng = np.random.default_rng(seed)
    unique_dates = np.unique(dates)

    # Row indices belonging to each date
    groups = {
        d: np.flatnonzero(dates == d)
        for d in unique_dates
    }

    base_rmse = np.sqrt(np.mean((y_true - pred_base) ** 2))
    alt_rmse = np.sqrt(np.mean((y_true - pred_alt) ** 2))
    observed_pct = 100 * (alt_rmse - base_rmse) / base_rmse

    boot_pct = []

    for _ in range(n_boot):
        sampled_dates = rng.choice(
            unique_dates, size=len(unique_dates), replace=True
        )
        idx = np.concatenate([groups[d] for d in sampled_dates])

        rmse_base = np.sqrt(np.mean((y_true[idx] - pred_base[idx]) ** 2))
        rmse_alt = np.sqrt(np.mean((y_true[idx] - pred_alt[idx]) ** 2))

        boot_pct.append(100 * (rmse_alt - rmse_base) / rmse_base)

    low, high = np.percentile(boot_pct, [2.5, 97.5])

    return {
        "RMSE change vs historical (%)": observed_pct,
        "95% bootstrap CI lower (%)": low,
        "95% bootstrap CI upper (%)": high,
        "Bootstrap replicates": n_boot,
        "Unique test dates": len(unique_dates)
    }

rows = []

for target, horizon in [
    ("target_combined_var_next1d", "Next day"),
    ("target_combined_var_next5d", "Next 5 trading observations")
]:
    dates, y, pred = fit_test_predictions(target)

    for alternative in [
        "Historical + all LLM",
        "Historical + focused LLM"
    ]:
        result = date_cluster_bootstrap(
            dates, y,
            pred["Historical-only"],
            pred[alternative],
            n_boot=2000,
            seed=42
        )
        rows.append({
            "Horizon": horizon,
            "Comparison": alternative + " vs historical-only",
            **result
        })

bootstrap_results = pd.DataFrame(rows)
display(bootstrap_results.round(4))

bootstrap_path = TABLES_DIR / "llm_rmse_date_cluster_bootstrap.csv"
bootstrap_results.to_csv(bootstrap_path, index=False)

print("Saved bootstrap results:", bootstrap_path)
print(
    "Interpretation: the RMSE change is positive when the LLM model is worse. "
    "If the confidence interval includes zero, the difference is not clearly "
    "distinguishable from zero under this bootstrap procedure."
)
```
